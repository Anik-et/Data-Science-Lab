SMOTE should be applied only to the training data after the train-test split. SMOTE generates synthetic minority-class examples by interpolating between existing minority samples. If SMOTE is applied before splitting, synthetic samples may contain information derived from observations that later end up in the test set. This introduces data leakage and produces overly optimistic evaluation metrics. By splitting first and applying SMOTE only on the training set, the test set remains an unbiased estimate of real-world performance.

The PCA components were generated after standardizing the original features, so they are already on comparable scales. Scaling them again is generally unnecessary. However, Amount has a much larger range than the PCA components, so I would scale Amount for algorithms that are sensitive to feature magnitude, such as Logistic Regression, SVM, or KNN.

Then I fit the scaler only on the training data so that the scaling parameters (such as mean and standard deviation for StandardScaler, or minimum and maximum values for MinMaxScaler) are learned only from the training distribution. I then use the same fitted scaler to transform both the training and test sets. This prevents information from the test set leaking into the training process.
